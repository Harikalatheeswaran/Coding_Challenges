### 1. Contains Duplicate : 

- [Link](https://neetcode.io/problems/duplicate-integer/question?list=neetcode150)
- `Time Taken` - `43`ms
- `Time Complexity` - `O(n)`
- `Space Complexity` - `O(n)`

````python
class Solution:
    def hasDuplicate(self, nums: List[int]) -> bool:
        if len(nums) == len(set(nums)):
            return False
        else: 
            return True
       

        
        
````

### 2. Valid Anagram : 

- [Link](https://neetcode.io/problems/is-anagram/question?list=neetcode150)
- `Time Taken` - `38`ms

- `Time Complexity` - `O(n^2)` (worst case, due to `count` inside loop over unique chars)
- `Space Complexity` - `O(n)`

````python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        # covering the lenght case
        if len(s) != len(t): 
            return False

        # Fucntion to convert string to numbered dictionaries
        def conv(s):
            out = {}
            for i in set(list(s)):
                out[i] = s.count(i)
            return out
        
        if conv(s) == conv(t): 
            return True
        
        return False

        

````

### 3. Two Sums : 

- [Link](https://neetcode.io/problems/two-integer-sum/question?list=neetcode150)
- `Time Taken` - `37`ms
- `Time Complexity` - `O(n^2)` (both solutions use `count`/`index`/`in` checks inside the main loop)
- `Space Complexity` - `O(n)`

````python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        diff_dict = {}
        out = []
        for i in range(len(nums)):
            # is to find if we have repeatitive ones & return that
            if nums[i] == target-nums[i] and nums.count(nums[i]) > 1:
                out.append(i)
                x = nums.pop(i) # removing the first occurence
                out.append(nums.index(x)+1)
                return out
            # to avoid a single element, when doubled adding to the target
            if nums[i] == target-nums[i] :#and nums.count(nums[i]) == 1:
                pass
            else:
                diff_dict[nums[i]] = target-nums[i], i
        
        for i,j in diff_dict.items():
            if diff_dict[i][0] in nums:
                out.append(j[1])
        return out
       
        
        
````

### 4. Group Anagrams : 

- [Link](https://neetcode.io/problems/anagram-groups/question?list=neetcode150)
- `Time Taken` - ``ms
- `Time Complexity` - 
- `Space Complexity` - 

````python
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        output = []
        temp = []

        # covering the null case
        if strs == [""] or strs == [" "]:
            return [[""]]
        
        # covering the single string case
        if len(strs) == 1:
            return [strs]

        # function to convert single string to a set of words
        def set_conversion(s):
            """Converts a string to a set of individual letters & we return the set with lenght"""
            return (s, len(s), set(list(s)))
        str_dict = {}
        
        # building a length based set
        for i in strs:
            t = set_conversion(i)
            if t[1] not in str_dict:
                str_dict[t[1]] = [t[2]]
            else:
                str_dict[t[1]].append(t[2])

        
            

        
            
        
````
